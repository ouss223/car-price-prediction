# Car Price Prediction — Business Understanding

## 1. Context
Used and new car prices in Tunisia vary widely across brands, models, age, condition, and seller channels. Buyers risk overpaying, sellers risk underpricing, and dealers need fast guidance to price and move inventory.

## 2. Problem Statement
We need a pricing assistant that gives a fair market price estimate for a car listing using its attributes (brand, model, year, mileage, body type, fuel, transmission, condition, location). The goal is to reduce pricing guesswork and speed up decisions for buyers, sellers, and dealers.

## 3. Business Objectives
- Provide price recommendations that are trusted and easy to act on.
- Shorten time-to-price for new listings.
- Reduce mispriced listings that sit too long or sell below value.

## 4. Stakeholders & Users
- Private sellers and buyers on marketplaces.
- Dealers managing inventory and trade-ins.
- Internal analysts monitoring market signals.



## 5. Key Questions
- Which attributes drive price most (brand/model, age, mileage, body type, fuel, transmission, condition)?
- How does price decay with age and mileage across segments?
- Do we need separate pricing for new vs. used cars?
- How stable are price drivers over recent periods?
